```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Load data
df = pd.read_csv('../data/water_data.csv')

# Show first 5 rows
df.head()
```

```python
# Group by community type
grouped = df.groupby('Community_Type')['Distance_km'].mean()
grouped
```

```python
# Plot chart
grouped.plot(kind='bar')
plt.title('Average Distance to Water by Community Type')
plt.xlabel('Community Type')
plt.ylabel('Distance (km)')
plt.tight_layout()

# Save chart
plt.savefig('../images/water_chart.png')
plt.show()
```
